```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize, wordpunct_tokenize, WhitespaceTokenizer
from nltk.stem import LancasterStemmer,PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag
```

```python
nltk.download('punkt_tab')
```

You can always refer to the [NLTK Documentation](https://www.nltk.org/)

```python
text = "Hi! I live in Groningen and have been here for over a year now. It's a very nice city; however, for the past two months I haven't been sure whether I actually like living here. The city itself is still nice, but the weather has been terrible lately. It's either snowing or raining, and in both cases walking and cycling are a struggle. I enjoy walking, and that's how I usually get to work, so now I don't really enjoy the commute anymore. I think I'll need to wait until April to start enjoying the city again. By the way, it's slippery outside, so be careful."
```

```python
text
```

# Tokenization

## Sentence Tokenization

Sentence tokenization by splitting on "."

```python
sentences_period = text.split(".")
for i, sent in enumerate(sentences_period):
  print(f"Sentence {i}:")
  print(sent)
```

Sentence tokenization using NLTK sent_tokenize

```python
sentences_nltk = sent_tokenize(text)
for i, sent in enumerate(sentences_nltk):
  print(f"Sentence {i}:")
  print(sent)
```

## Word Tokenizaton

Word tokenization by splitting on white space

```python
words_ws = text.split()
for i, word in enumerate(words_ws):
  print(f"Word {i}: {word}")
```

Word tokenization using NLTK word_tokenize

```python
words_nltk = word_tokenize(text)
for i, word in enumerate(words_nltk):
  print(f"Word {i}: {word}")
```

Word tokenization using NLTK wordpunct_tokenize

```python
words_p_nltk = wordpunct_tokenize(text)
for i, word in enumerate(words_p_nltk):
  print(f"Word {i}: {word}")
```

## Stemming & Lemmatization

### Stemming

```python
porter_stemmer = PorterStemmer()
lancaster_stemmer = LancasterStemmer()
```

```python
words = ["connect", "connected", "connecting", "connection"]
for word in words:
  print(f"Stem of {word} is {porter_stemmer.stem(word)}")
```

```python
words = ["university", "universal", "universe"]
for word in words:
  print(f"Stem of {word} is {porter_stemmer.stem(word)}")
```

```python
word_counts = Counter(words_nltk)
print(f"Number of unique tokens in 'text': {len(word_counts)}")
```

```python
word_pstemms = [porter_stemmer.stem(word) for word in words_nltk]
word_pstemms_counts = Counter(word_pstemms)
print(f"Number of unique tokens after stemming with PorterStemmer: {len(word_pstemms_counts)}")
```

```python
word_lstemms = [lancaster_stemmer.stem(word) for word in words_nltk]
word_lstemms_counts = Counter(word_lstemms)
print(f"Number of unique tokens after stemming with LancasterStemmer: {len(word_lstemms_counts)}")
```

### Lemmatiziation

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize(text)
```

```python
for word in words_nltk:
  print(f"Lemma of {word}: {lemmatizer.lemmatize(word)}")
```

```python
words = ["live", "living", "lives", "life", "lived"]
for word in words:
  print(f"Lemma of {word}: {lemmatizer.lemmatize(word)}")
```

```python
words = ["live", "living", "lives", "life", "lived"]
for word in words:
  print(f"Lemma of {word}: {lemmatizer.lemmatize(word, wordnet.ADJ)}")
```

# Parts-Of-Speech Tagging

```python
tags = pos_tag(words_nltk)
```

```python
tags
```

```python
def wn_pos(tag):
  if tag.startswith('J'):
        return wordnet.ADJ
  if tag.startswith('V'):
      return wordnet.VERB
  if tag.startswith('N'):
      return wordnet.NOUN
  if tag.startswith('R'):
      return wordnet.ADV
  return wordnet.NOUN
```

```python
for i, word in enumerate(words_nltk):
  print(f"Lemma of {word}: {lemmatizer.lemmatize(word, wn_pos(tags[i][1]))}")
```
