#Named Entity Recognition

## Using NLTK

You can always refer to the [NLTK Documentation](https://www.nltk.org/)

```python
import nltk
from nltk import word_tokenize, pos_tag
from nltk.chunk import ne_chunk
from nltk.tree import Tree
```

```python
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("maxent_ne_chunker_tab")
nltk.download("words")
```

```python
tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)
tree = ne_chunk(pos_tags)
```

```python
tree
```

```python
def extract_nltk_entities(chunk_tree):
    entities = []
    for node in chunk_tree:
        if isinstance(node, Tree):
            label = node.label()
            entity_text = " ".join(token for token, pos in node.leaves())
            entities.append((entity_text, label))
    return entities
```

```python
nltk_entities = extract_nltk_entities(tree)
nltk_entities
```

## Using spaCy

You can always refer to [spaCy documentation](https://spacy.io/usage/linguistic-features#named-entities)

```python
import spacy
from spacy import displacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(text)

[(ent.text, ent.label_) for ent in doc.ents]
```

```python
displacy.render(doc, style="ent", jupyter=True)
```

```python
for ent in doc.ents:
    print(f"{ent.text:<25} {ent.label_:<10} start={ent.start_char} end={ent.end_char}")
```

## NLTK vs spaCy

```python
spacy_entities = [(ent.text, ent.label_) for ent in doc.ents]

print("NLTK entities:")
for e in nltk_entities:
    print("  ", e)

print("\nspaCy entities:")
for e in spacy_entities:
    print("  ", e)
```

# Performance

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
```

## Example of a balanced dataset

```python
y_true = np.array([0,0,0,0,1,1,1,2,2,2])
y_pred = np.array([0,0,1,0,1,2,1,2,0,2])

print("True labels:     ", y_true)
print("Predicted labels:", y_pred)
```

```python
cm = confusion_matrix(y_true, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
plt.show()
```

```python
acc = accuracy_score(y_true, y_pred)
print("Accuracy:", acc)
```

```python
print("Precision (macro):  ", precision_score(y_true, y_pred, average="macro"))
print("Recall (macro):     ", recall_score(y_true, y_pred, average="macro"))
print("F1 (macro):         ", f1_score(y_true, y_pred, average="macro"))

print("Precision (micro):  ", precision_score(y_true, y_pred, average="micro"))
print("Recall (micro):     ", recall_score(y_true, y_pred, average="micro"))
print("F1 (micro):         ", f1_score(y_true, y_pred, average="micro"))

print("F1 (weighted):      ", f1_score(y_true, y_pred, average="weighted"))
```

```python
print(classification_report(y_true, y_pred))
```

## Example of an unbalanced dataset

```python
y_true_imb = np.array([0]*90 + [1]*7 + [2]*3)
y_pred_imb = np.array([0]*95 + [1]*3 + [2]*2)
```

```python
cm_imb = confusion_matrix(y_true_imb, y_pred_imb)

disp = ConfusionMatrixDisplay(confusion_matrix=cm_imb)
disp.plot()
plt.show()
```

```python
print("Accuracy:", accuracy_score(y_true_imb, y_pred_imb))
print("F1 macro:", f1_score(y_true_imb, y_pred_imb, average="macro"))
print("F1 micro:", f1_score(y_true_imb, y_pred_imb, average="micro"))
print("F1 weighted:", f1_score(y_true_imb, y_pred_imb, average="weighted"))

print("Detailed report:")
print(classification_report(y_true_imb, y_pred_imb))
```
