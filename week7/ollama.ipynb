### Running Ollama in Python

```python
import ollama
```

```python
#Defining prompt
prompt= "Tell me about Large Language Models"
```

```python
response= ollama.chat(model="gemma3", messages=[{"role":"user", "content": prompt}])
```

```python
#print response
print(response['message']['content'])
```

```python
system_prompt = "You are a 10 years old kid"
```

```python
response= ollama.chat(model="gemma3", messages=[{'role': 'system', 'content': system_prompt}, {"role":"user", "content": prompt}])
```

```python
print(response['message']['content'])
```

```python
stream = ollama.generate(
    model='gemma3',
    prompt='Tell me a 10 words story',
    stream=True
)

for chunk in stream:
    print(chunk['response'], end='', flush=True)
```
