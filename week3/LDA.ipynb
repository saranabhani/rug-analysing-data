# Importing the required libraries

```python
import os
import pandas as pd
import numpy as np
import glob
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
nlp.max_length = 4000000
```

# Loading the text data

```python
def read_files(files):
  corpus = []
  filenames = []
  for file in files:
    with open(file, 'r', encoding='utf-8') as f:
      corpus.append(f.read())
      filenames.append(file.split("/")[-1][:-4])
  return corpus, filenames
```

```python
files = glob.glob("week3/books/*.txt")
```

```python
corpus, filenames = read_files(files)
```

```python
print(f"Number of documents: {len(corpus)}")
```

# Topic Modeling 1

## Vectorizing the text data

* Note that for LDA, we use term frequencies (CountVectorizer and not TfidfVectorizer) because LDA assumes each document is a bag of words counts and the words are drawn from a discrete probability distribution. while TF-IDF gives the weights of the words.

```python
tf_vectorizer = CountVectorizer(encoding='utf-8', lowercase=True, max_df=0.95, min_df=3, max_features=2500)
corpus_vectorized = tf_vectorizer.fit_transform(corpus)
```

```python
tf_feature_names = tf_vectorizer.get_feature_names_out()
```

```python
tf_feature_names[:10]
```

```python
tf_feature_names[-10:]
```

```python
len(tf_feature_names)
```

## Training the LDA model

```python
lda = LatentDirichletAllocation(n_components=4, random_state=7, learning_method='online')
lda.fit(corpus_vectorized)
```

## Visualizing the topics

```python
def plot_top_words(model, feature_names, n_top_words, title):
    fig, axes = plt.subplots(1, 4, figsize=(30, 15), sharex=True)
    axes = axes.flatten()
    for topic_idx, topic in enumerate(model.components_):
        top_features_ind = topic.argsort()[-n_top_words:]
        top_features = feature_names[top_features_ind]
        weights = topic[top_features_ind]

        ax = axes[topic_idx]
        ax.barh(top_features, weights, height=0.7)
        ax.set_title(f"Topic {topic_idx +1}", fontdict={"fontsize": 30})
        ax.tick_params(axis="both", which="major", labelsize=20)
        for i in "top right left".split():
            ax.spines[i].set_visible(False)
        fig.suptitle(title, fontsize=40)

    plt.subplots_adjust(top=0.90, bottom=0.05, wspace=0.90, hspace=0.3)
    plt.show()
```

```python
n_top_words = 30
plot_top_words(lda, tf_feature_names, n_top_words, 'Topics in LDA model')
```

```python
# document topic distribution
doc_topic_dist = lda.transform(corpus_vectorized)
df = pd.DataFrame(doc_topic_dist, columns=["Topic 1", "Topic 2", "Topic 3", "Topic 4"])
df['filename'] = filenames
print(df)
```

# Topic Modeling 2: Removing stopwords

## Preprocessing the text data

```python
# remove stop words
def remove_sw(text):
    doc = nlp(text)
    filtered_text = [token.lower_ for token in doc if not token.is_stop]
    return ' '.join(filtered_text)
```

```python
corpus_preprocessed = [remove_sw(text) for text in corpus]
```

## Vectorizing the preprocessed text data

```python
corpus_vectorized = tf_vectorizer.fit_transform(corpus_preprocessed)
```

```python
tf_feature_names = tf_vectorizer.get_feature_names_out()
```

## Training the LDA model

```python
lda.fit(corpus_vectorized)
```

```python
plot_top_words(lda, tf_feature_names, n_top_words, 'Topics in LDA model')
```

```python
# document topic distribution
doc_topic_dist = lda.transform(corpus_vectorized)
df = pd.DataFrame(doc_topic_dist, columns=["Topic 1", "Topic 2", "Topic 3", "Topic 4"])
df['filename'] = filenames
print(df)
```

# Topic Modeling 4: + Entities Masking

## Preprocessing the text data

```python
# remove stop words + lemmatize + mask entities
def mask_lemma_remove_sw(text):
    doc = nlp(text)
    filtered_text = []
    for token in doc:
        if not token.is_stop and not token.is_punct and not token.is_space and token.is_alpha:
            if token.ent_type_ in ['PERSON', 'ORG', 'GPE'] or token.pos_ in ['PROPN', 'NUM', 'SYM']:
                filtered_text.append('MASKED')
            elif token.pos_ in ['DET', 'ADP', 'CCONJ', 'PRON', 'AUX', 'PART', 'PUNCT', 'INTJ']:
                continue
            else:
                filtered_text.append(token.lemma_.lower())
    return ' '.join(filtered_text)
```

```python
corpus_preprocessed = [mask_lemma_remove_sw(text) for text in corpus]
```

## Vectorizing the preprocessed text data

```python
corpus_vectorized = tf_vectorizer.fit_transform(corpus_preprocessed)
```

```python
tf_feature_names = tf_vectorizer.get_feature_names_out()
```

## Training the LDA model

```python
lda.fit(corpus_vectorized)
```

```python
plot_top_words(lda, tf_feature_names, n_top_words, 'Topics in LDA model')
```

```python
# document topic distribution
doc_topic_dist = lda.transform(corpus_vectorized)
df = pd.DataFrame(doc_topic_dist, columns=["Topic 1", "Topic 2", "Topic 3", "Topic 4"])
df['filename'] = filenames
print(df)
```
